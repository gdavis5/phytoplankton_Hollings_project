# Satellite Oceanography: Development of Phytoplankton Indicators from Ocean Color Remote Sensing

### Phytoplankton on the NES
Phytoplankton are autotrophic plankton that serve as the base of much of the marine food web. Growth of phytoplankton is governed by two key variables, light availability and nutrient availability. In the tropical and subtropical latitudes, phytoplankton growth is limited by nutrients, since there is an abundance of light but warmer, less nutrient dense water. In polar and mid-latitudes, phytoplankton growth is mainly limited by light availability (Fischer et al 2014). Studying phytoplankton growth and phenology is vital to understanding ecosystems and changes to those ecosystems over time. Phytoplankton serve as food for zooplankton, such as fish larvae, so out-of-sync growth periods can create an overabundance of phytoplankton when zooplankton are not feeding and very limited supply of phytoplankton when they are (Cushing 1969). In highly biologically productive areas, such as the Northeast United States Continental Shelf, understanding phytoplankton blooms and bloom phenology is vital for managing marine resources. This region usually has two major phytoplankton blooms throughout the year, one in the spring and a smaller one in the fall (Townsend et al. 2004). The region can be further divided into 3 regions: the Gulf of Maine, Georges Bank, and the Middle Atlantic Bight. These regions each see their own phytoplankton bloom characteristics and provide important ecosystems for local marine species that drive fisheries. For this project, we use five regions: Middle Atlantic Bight South, Middle Atlantic Bight North, Georges Bank, Gulf of Maine West, and Gulf of Maine East (Figure #). 

### Satellite Oceanography
Satellites are very useful for studying phytoplankton blooms and their phenology because they remotely track and quantify physical and optical oceanographic properties. Studying phytoplankton from satellites requires sensors that measure ocean spectral radiance. These sensors measure wavelengths of light emitted from the ocean subsurface which are excellent proxies for the amount of phytoplankton or particles in the water. As phytoplankton concentrations increase, less blue light is scattered and more green light is scattered since chlorophyll-a is the main photosyntheic pigment in phtyoplankton (Werdell and McClain 2019). We can use the spectral radiance to quantify chlorophyll-a concentrations, which allows us to study phytoplankton blooms (IOCCG report Number 7, 2008, Chapter 6). Using satellites provides a great temporal and spatial advantage to study phytoplankton, allowing for essentially global, 24 hour coverage of the oceans by using a network of satellites. The disadvantage to satellite data is that it can contain missing data and gaps because of cloud interference, temporal resolution, and sensor miscalibration. Despite this disadvantage, satellites serve as one of the best ways to study large-scale and regional phenomena due to higher resolution global converage. Ocean color products are vital in studying surface phytoplankton at the regional and global scales. To ensure daily coverage, a variety of satellites in a network are used to validate and fill in missing data, one such project that utilizes ocean color and satellite networks is the Ocean Colour Climate Change Initiative (OCCCI) project.

### OCCCI Chl-a products

OCCCI, a European Space Agency (ESA) sponsored product, observes water-leaving radiance in the visible spectrum to derive chlorophyll-a concentrations in the surface layer of the ocean. The version 6.0 dataset, released in 2022, contains a variety of ocean color data, focusing on regions where optical properties are dominated by phytoplankton (ESA 2024). The data is a compilation from the Medium Resolution Imaging Spectrometer (MERIS), Moderate Resolution Imaging Spectroradiometer (MODIS), Visible Infrared Imaging Radiometer Suite (VIIRS), Sea-Viewing Wide Field-of-View Sensor (SeaWiFS), Global Area Coverage (GAC), Sentinel 3a and 3b Ocean and Land Colour Instrument (OLCI) satellites (ESA 2024). Using data from these satellites, the OCI, OCI2, OC2, and OCx algorithms were merged together to create the chlorophyll products in OCCCI (ESA 2024). OCCCI provides daily, 5-day, 8-day, monthly, and annual composites of a variety of chlorophyll metrics, including median chlorophyll data, which is the primary value of chlorophyll-a used in this project (ESA). We chose median chlorophyll-a because the mean chlorophyll-a is likely very skewed by the presence of very extreme phytoplankton blooms so the median is likely a better statistic to observe and it commonly used in other phytoplankton studies. The OCCCI project provides a time series beginning in September 1997 and is updated quarterly, so the most recent data is through the end of September 2025. Using OCCCI data will make it easy to continue to update the data used in this project, allowing it to serve as a way for continuous study of phytoplankton phenology on the NES. 

### Current state of NES phytoplankton phenology studies and where ours fits

For this project, there are two main objectives: to select a bloom detection method that most accurately models phytoplankton blooms along the NES and to quantify a variety of bloom metrics to determine time series trends useful for fisheries management. These objectives aim to quantify blooms on the NES and bloom metrics to accurate track changes to long-term phytoplankton bloom trends for further use in fisheries recommendations. We aim to develop a model to continue to use OCCCI data to model surface chlorophyll-a to identify long-term trends in phytoplankton bloom phenology. 

#### Bloom Detection Methods
After assessing a variety of phytoplankton phenology literature, several bloom detection methods were found, including threshold based methods, rate of change based methods, and maximum concentration methods. For the scope of the project, we have chosen to focus on testing a threshold-based method and the rate of change method. They are two of the most common bloom detection methods and they come recommended to study phenology changes and seasonal physical and biological mechanisms that create blooms, which is the other main focus of this project (Brody, Lozier, and Dunne 2013). The threshold method is done by calculating the climatological median for the area and then finding a certain percentage of that median and find points above that percentage. Based on the literature, the threshold percentage likely falls between 5-30% above the climatological median for most phenology studies, so we are initially testing within this range in 5% intervals to determine a value that accurately represents the NES. The rate of change method focuses on finding the daily rate of change of chlorophyll-a concentration and finding the maximum rate of change prior to a peak. This date of maximum rate of change is considered the bloom initiation date.  

#### Bloom Metrics
After defining a phytoplankton bloom for our region, we will then quantify different bloom metrics that can be used to investigate changes to phytoplankton phenology. We initially focus on quantifying the number of bloom days and bloom events per year. We will get a count for each year in the time series (1998-2025) and then compare that to the climatological (Median/mean?). This will tell us if events are becoming more or less frequent each year. We will also analyze bloom initiation and termination dates and use those to determine if the timing of the blooms are shifting and if the duration is increasing, decreasing, or remaining the same compared to the climatological (mean/median?). We also hope to quantify the total yearly phytoplankton biomass by integrating chlorophyll-a concentrations each year to determine if the blooms are changing in severity. Another important metric for fisheries management is spatial shifts in phytoplankton blooms. To investigate this, we will define the center of gravity of the major spring and fall blooms and see if that has changed over the time series. These metrics, when compared to the regional climatology, should help in identifying temporal and spatial changes to bloom phenology on the NES.
