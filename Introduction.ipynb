### Phytoplankton on the NES
Phytoplankton are autotrophic plankton that serve as the base of much of the marine food web. Growth of phytoplankton is governed by two key variables, light availability and nutrient availability. In the tropical and subtropical latitudes, phytoplankton growth is limited by nutrients, since there is an abundance of light but warmer, less nutrient dense water. In polar and mid-latitudes, phytoplankton growth is mainly limited by light availability (Fischer et al 2014). Studying phytoplankton growth and phenology is vital to understanding ecosystems and changes to those ecosystems over time. Phytoplankton serve as food for zooplankton, such as fish larvae, so out-of-sync growth periods can create an overabundance of phytoplankton when zooplankton are not feeding and very limited supply of phytoplankton when they are (Cushing 1969). In highly biologically productive areas, such as the Northeast United States Continental Shelf, understanding phytoplankton blooms and bloom phenology is vital for managing marine resources. This region usually has two major phytoplankton blooms throughout the year, one in the spring and a smaller one in the fall (Townsend et al. 2004). The region can be further divided into 3 regions: the Gulf of Maine, Georges Bank, and the Middle Atlantic Bight. These regions each see their own phytoplankton bloom characteristics and provide important ecosystems for local marine species that drive fisheries. 

### Satellite Oceanography
Satellites are very useful for studying phytoplankton blooms and their phenology because they remotely track and quantify physical and optical oceanographic properties. Studying phytoplankton from satellites requires sensors that measure ocean spectral radiance. These sensors measure wavelengths of light emitted from the ocean subsurface which are excellent proxies for the amount of phytoplankton or particles in the water. As phytoplankton concentrations increase, less blue light is scattered and more green light is scattered since chlorophyll-a is the main photosyntheic pigment in phtyoplankton (Werdell and McClain 2019). We can use the spectral radiance to quantify chlorophyll-a concentrations, which allows us to study phytoplankton blooms (IOCCG report Number 7, 2008, Chapter 6). Using satellites provides a great temporal and spatial advantage to study phytoplankton, allowing for essentially global, 24 hour coverage of the oceans by using a network of satellites. The disadvantage to satellite data is that it can contain missing data and gaps because of cloud interference, temporal resolution, and sensor miscalibration.

### Project Focus
For this project, there are two main objectives: to select a bloom detection method that most accurately models phytoplankton blooms along the NES and to quantify a variety of bloom metrics to determine time series trends useful for fisheries management. These objectives aim to quantify blooms on the NES and bloom metrics to accurate track changes to long-term phytoplankton bloom trends for further use in fisheries recommendations. 
#### Bloom Detection Methods
After assessing a variety of phytoplankton phenology literature, several bloom detection methods were found, including threshold based methods, rate of change based methods, and maximum concentration methods. For the scope of the project, we have chosen to focus on testing a threshold-based method and the rate of change method. They are two of the most common bloom detection methods and they come recommended to study phenology changes and seasonal physical and biological mechanisms that create blooms, which is the other main focus of this project (Brody, Lozier, and Dunne 2013). The threshold method is done by calculating the climatological median for the area and then finding a certain percentage of that median and find points above that percentage. Based on the literature, the threshold percentage likely falls between 5-30% above the climatological median for most phenology studies, so we are initially testing within this range in 5% intervals to determine a value that accurately represents the NES. The rate of change method focuses on finding the daily rate of change of chlorophyll-a concentration and finding the maximum rate of change prior to a peak. This date of maximum rate of change is considered the bloom initiation date.  
#### Bloom Metrics
After defining a phytoplankton bloom for our region, we will then work on quantifying different bloom metrics that can be used to investigate changes to phytoplankton phenology. We intend to focus on:
* Number of phytoplankton bloom days per year
* Number of bloom events per year
* Bloom initiation and termination date and the duration of blooms
* Center of gravity of blooms
* Integrated chlorophyll-a concentrations
* Periodicity of blooms

These metrics, when compared to the regional climatology, should help is identify temporal and spatial changes to bloom phenology on the NES. We then hope to examine particularly anomalous years and see what physical phenomena might have been driving the anomalous behavior.
