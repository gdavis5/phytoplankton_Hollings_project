# Satellite Oceanography: Development of Phytoplankton Indicators from Ocean Color Remote Sensing

### Phytoplankton 
Phytoplankton are autotrophic plankton that serve as the base of much of the marine food web. Growth of phytoplankton is governed by two key ingredients, light availability and nutrient availability. In the tropical and subtropical latitudes, phytoplankton growth is limited by nutrients, since there is an abundance of light but warmer, less nutrient dense water. In polar and mid-latitudes, phytoplankton growth is mainly limited by light availability (Fischer et al 2014). Studying phytoplankton growth and phenology is vital to understanding ecosystems and changes to those ecosystems over time. Phytoplankton serve as food for zooplankton, such as fish larvae, so out-of-sync growth periods can create an overabundance of phytoplankton when zooplankton are not feeding and very limited supply of phytoplankton when they are (Cushing 1969).

### Satellite Oceanography
Satellite oceanography is a branch of oceanography concerned with using satellites to remotely track and quantify oceanographic properties and optical properties. Sensors use metrics they can quantitatively measure, such as ocean color and ocean spectral reflectance, to estimate other oceanographic variables, such as chlorophyll a concentration. Using satellites provides a great temporal and spatial advantage, as we can gather measurements over a larger area in a shorter amount of time compared to in situ measurements. This allows us to have essentially global, 24 hour coverage of the oceans by using a variety of satellites. The disadvantage to satellite data is that it can contain missing data and gaps because of cloud interference, temporal resolution, and sensor miscalibration. Despite this disadvantage, satellite data provides us with information about chlorophyll-a concentrations in the surface waters, which serves as an excellent proxy measurement for phytoplankton growth (IOCCG report Number 7, 2008, Chapter 6). We use satellite data to study regional phytoplankton phenology to understand the regional ecosystems.

### Northeast US Contential Shelf
For this study, we are focused on the Northeast United States Continental Shelf (NES). Located in the Northwest Atlantic Ocean, this region has a very dynamic ecosystem characterized by high biological productivity. This region usually has two major phytoplankton blooms throughout the year, one in the spring and a smaller one in the fall (Townsend et al. 2004). The region can be divided into 3 regions: the Gulf of Maine, Georges Bank, and the Middle Atlantic Bight.

### Project Focus
For this project, there are two main objectives: to select a bloom detection method that most accurately models phytoplankton blooms along the NES and to quantify a variety of bloom metrics to determine time series trends useful for fisheries management. These objectives aim to quantify blooms on the NES and bloom metrics to accurate track changes to long-term phytoplankton bloom trends for further use in fisheries recommendations. 
#### Bloom Detection Methods
After assessing a variety of phytoplankton phenology literature, several bloom detection methods were found, including threshold based methods, rate of change based methods, and maximum concentration methods. For the scope of the project, we have chosen to focus on testing a threshold-based method and the rate of change method. They are two of the most common bloom detection methods and they come recommended to study phenology changes and seasonal physical and biological mechanisms that create blooms, which is the other main focus of this project (Brody, Lozier, and Dunne 2013).
#### Bloom Metrics
After defining a phytoplankton bloom for our region, we will then work on quantifying different bloom metrics that can be used to investigate changes to phytoplankton phenology. We intend to focus on:
* Number of phytoplankton bloom days per year
* Number of bloom events per year
* Duration of blooms
* Center of gravity of blooms
* Integrated chlorophyll-a concentrations
* Bloom initiation date
* Periodicity of blooms
* Minimum, maximum, and mean chlorophyll-a concentrations

These metrics, when compared to the regional climatology, should help is identify temporal and spatial changes to bloom phenology on the NES. We then hope to examine particularly anomalous years and see what physical phenomena might have been driving the anomalous behavior.
